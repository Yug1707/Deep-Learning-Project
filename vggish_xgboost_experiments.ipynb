# VGGish + XGBoost Experiments\n
\n
This notebook runs and inspects the parallel VGGish + XGBoost pipeline without changing the existing CRNN workflow.

```python
from pathlib import Path\n
import json\n
import subprocess\n
import sys\n
\n
ROOT = Path.cwd()\n
CONFIG = ROOT / 'training' / 'vggish_xgboost' / 'config_vggish_xgboost.json'\n
CONFIG
```

```python
# Optional: install additive dependencies for this pipeline\n
# !{sys.executable} -m pip install -r requirements_vggish.txt
```

```python
# 1) Build balanced manifests\n
subprocess.run([sys.executable, 'training/vggish_xgboost/build_dataset.py', '--config', str(CONFIG)], check=True)
```

```python
# 2) Extract VGGish embeddings\n
subprocess.run([sys.executable, 'training/vggish_xgboost/extract_embeddings.py', '--config', str(CONFIG)], check=True)
```

```python
# 3) Train One-vs-Rest XGBoost\n
subprocess.run([sys.executable, 'training/vggish_xgboost/train_xgboost.py', '--config', str(CONFIG)], check=True)
```

```python
# 4) Evaluate on val/test\n
subprocess.run([sys.executable, 'training/vggish_xgboost/evaluate_xgboost.py', '--config', str(CONFIG)], check=True)
```

```python
report_path = ROOT / 'logs' / 'vggish_xgboost' / 'results' / 'evaluation_report.json'\n
with open(report_path, 'r', encoding='utf-8') as f:\n
    report = json.load(f)\n
report
```
